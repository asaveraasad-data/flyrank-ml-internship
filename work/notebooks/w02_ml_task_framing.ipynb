## 1. My lane as an ML task (type)

I chose the **Content Refresh Prioritization** lane.

This is primarily a **ranking/scoring** problem.

The goal is not simply to predict whether a page is declining. Instead, the goal is to help content editors decide **which pages should be reviewed first**. Every page receives a priority score, and pages with the highest scores are reviewed before others.

This supports a real business decision by helping teams spend their limited time on the pages most likely to benefit from a refresh.

```python
print("ML Task Type: Ranking / Scoring")
print("Lane: Content Refresh Prioritization")
```

## 2. Target or proxy

The output is a **priority score** that ranks pages for review.

For learning purposes, the starter dataset provides an observed label indicating whether a page is declining.

The target is derived from observed performance trends in the dataset. During modeling, observable metrics such as impressions, CTR, engagement, position, and content freshness can be used, while variables that directly define the label (such as trend_direction or trend_pct) should not be used as model features.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

df["is_declining_label"] = (df["trend_direction"] == "down").astype(int)

print(df["is_declining_label"].value_counts())
```

## 3. Success metric

The primary success metric is **Precision@50**.

Precision@50 measures how many of the top 50 recommended pages are actually declining.

This metric matches the real business objective because editors usually review only a limited number of pages, so ranking the most important pages near the top is more valuable than maximizing overall accuracy.

```python
print("Success Metric: Precision@50")
```

## 4. The unit of analysis

Each row represents **one content page**.

Each page contains search performance, engagement metrics, content characteristics, freshness information, and trend indicators.

The model makes one recommendation for each individual page.

```python
print("Dataset Shape:", df.shape)

df.head()
```

## 5. Why ML beats a fixed rule here

A simple rule such as:

"If impressions are high and the page has not been updated recently, recommend it."

is useful but cannot capture the complex relationships between many search signals.

Different pages may have similar impressions but very different CTR, engagement, rankings, competition levels, or content age.

Machine learning can learn these patterns from historical observations and combine many signals simultaneously to produce better rankings than a single fixed rule.

The output should be used as decision support rather than as proof of future outcomes.

```python
print("Example signals available for ML:")

features = [
    "impressions_90d",
    "ctr",
    "avg_position",
    "engagement_rate",
    "content_age_days"
]

print(df[features].describe())
```
