## Research Question

Decision:
Which content pages should be reviewed first?

Who acts?
An SEO specialist or content editor.

Action:
Review and update the highest-priority pages before spending time on lower-impact pages.

Cost of a wrong recommendation:
- False Positive: An editor wastes time reviewing a page that didn't need updating.
- False Negative: A declining page is ignored, causing continued traffic loss.

Why ML?
Simple rules based on one metric (such as impressions or content age) cannot capture the complex relationships between many search-performance signals.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Rows:", len(df))
print("Columns:", len(df.columns))

corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Correlation between search_volume and impressions_90d: {corr:.3f}")

wc = df.groupby("trend_direction")["word_count"].median()
print("\nMedian word count by trend direction:")
print(wc)
```

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))

corr = df["search_volume"].corr(df["impressions_90d"])
print(f"\nCorrelation between search_volume and impressions_90d: {corr:.3f}")

median_words = df.groupby("trend_direction")["word_count"].median()

print("\nMedian Word Count")
print(median_words)

print("\nAverage CTR by Position Tier")

ctr = df.groupby("position_tier")["ctr"].mean().sort_values(ascending=False)

print(ctr)
```

### Observations

- The starter dataset contains **30,000 pages** and **44 features**.
- The correlation between search volume and impressions is **approximately 0.001**, suggesting a very weak relationship in this dataset.
- Pages appearing higher in search results generally have higher average CTR.
- Median word counts for growing and declining pages are similar, suggesting that content length alone is not a strong indicator of page performance.

These observations suggest that multiple signals together are likely more informative than relying on a single rule.
